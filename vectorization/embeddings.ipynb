## Embeddings and Vector Store

```python
from langchain_huggingface import HuggingFaceEmbeddings

# Initialize the HuggingFace embeddings model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

embeddings
```

```python
text="Napoleon Bonaparte was a French military leader and emperor who rose to prominence during the French Revolution and its associated wars. He is best known for his role in the Napoleonic Wars, where he led France against various European coalitions. Napoleon implemented numerous reforms across Europe, including the Napoleonic Code, which influenced legal systems worldwide. Despite his military genius, his ambitions ultimately led to his downfall, culminating in his defeat at the Battle of Waterloo in 1815 and his subsequent exile to the island of Saint Helena, where he died in 1821."
embedding = embeddings.embed_query(text)
print("Embedding vector:", embedding)
print("Embedding vector length:", len(embedding))
print("First 5 dimensions of the embedding vector:", embedding[:5])
```

```python
sentences = [
    "The Eiffel Tower is located in Paris.",
    "The Great Wall of China is visible from space.",
    "Mount Everest is the highest mountain in the world."
]

sentence_embeddings = embeddings.embed_documents(sentences)
for i, sentence in enumerate(sentences):
    print(f"\nSentence: {sentence}")
    print(f"Embedding vector: {sentence_embeddings[i]}")
    print(f"Embedding vector length: {len(sentence_embeddings[i])}")
    print(f"First 5 dimensions of the embedding vector: {sentence_embeddings[i][:5]}")  
```

## Cosine Similarity Example

```python
sentences = [
    "The Eiffel Tower is located in Paris.",
    "The Great Wall of China is visible from space.",
    "Mount Everest is the highest mountain in the world.",
    "The Nile is the longest river in the world.",
    "The Amazon rainforest is the largest tropical rainforest.",
    "The Sahara is the largest hot desert on earth.",
    "There is a mouse on the table."
]
```

```python
import numpy as np

def cosine_similarity(vec1, vec2):
    dot_product = sum(a * b for a, b in zip(vec1, vec2))
    magnitude1 = sum(a ** 2 for a in vec1) ** 0.5
    magnitude2 = sum(b ** 2 for b in vec2) ** 0.5
    if magnitude1 == 0 or magnitude2 == 0:
        return 0
    return dot_product / (magnitude1 * magnitude2)
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")   
embeddings 
```

```python
sentence_embedding = embeddings.embed_documents(sentences)
sentence_embedding
```

```python
# Calculate cosine similarities
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        sim = cosine_similarity(sentence_embedding[i], sentence_embedding[j])
        print(f"Cosine similarity between:\n'{sentences[i]}'\nand\n'{sentences[j]}'\nis: {sim:.4f}\n")
```

```python
# Semanic Search - find most relevant document to a query
document_texts = [
    "Ganymede is the largest moon of Jupiter and the largest moon in the solar system.",
    "Callisto is the second-largest moon of Jupiter and the third-largest moon in the solar system.",
    "Io is the innermost of the four Galilean moons of Jupiter and is the most volcanically active body in the solar system.",
    "Europa is the smallest of the four Galilean moons of Jupiter and is believed to have a subsurface ocean beneath its icy crust.",
    "I want a turkey sandwich with lettuce and tomato."
]

query_text = "Which moon of Jupiter is the largest?"

```

```python
def semantic_search(query, documents, embeddings, top_k=3):
    query_embedding = embeddings.embed_query(query)
    document_embeddings = embeddings.embed_documents(documents)
    
    similarities = []
    for i, doc_embedding in enumerate(document_embeddings):
        sim = cosine_similarity(query_embedding, doc_embedding)
        similarities.append((i, sim))
    
    similarities.sort(key=lambda x: x[1], reverse=True)
    
    top_documents = [(documents[i], sim) for i, sim in similarities[:top_k]]
    return top_documents
```

```python
semantic_search_results = semantic_search(query_text, document_texts, embeddings, top_k=3)
print("Top relevant documents for the query:")
for doc, score in semantic_search_results:
    print(f"Document: {doc}\nSimilarity Score: {score:.4f}\n")
```

```python
query_text = "Which moon of Jupiter is the tiniest?"

semantic_search_results = semantic_search(query_text, document_texts, embeddings, top_k=3)
print("Top relevant documents for the query:")
for doc, score in semantic_search_results:
    print(f"Document: {doc}\nSimilarity Score: {score:.4f}\n")  
```
